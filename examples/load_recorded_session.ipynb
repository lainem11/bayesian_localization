```python
from ABL import LocalizationSession
import os
from pathlib import Path

def _find_repo_root():
    for p in [Path.cwd(), *Path.cwd().parents]:
        if (p / 'pyproject.toml').exists():
            return str(p)
    return str(Path.cwd())

wrkdir = _find_repo_root()
```

```python
# Locate the session directory
example_session = os.path.join(wrkdir, "examples", "saved", "example_subject", "example_session")

# Start localization instance
S = LocalizationSession.load_experiment(example_session)
```

```python
# Save created figure
S.visualizer.figure.write_image(os.path.join(wrkdir, "images", "localization_example.png"), scale=2.0)
```

```python
S.end_session()
```
